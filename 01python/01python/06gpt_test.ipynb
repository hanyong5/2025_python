```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()


client = OpenAI(
    api_key=os.getenv('OPENAI_API_KEY')
)

# response = client.responses.create(
#     model="gpt-3.5-turbo",
#     input="서울에서 주말에 갈 만한 곳 추천해줘"
# )
response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {
            "role":"system",
            "content":(
                '''You are a professional personal trainer with expertise in exercise physiology.

Based on the user's input, calculate the number of calories burned during running using the METs (Metabolic Equivalent of Task) formula. The user will provide the following information:

- Running time (minutes)  
- Distance (km)  
- Body weight (kg)  
- Age  
- Gender  

Please follow these steps:

1. Estimate the running speed (km/h) from the distance and time, and determine the appropriate METs value.
2. Explain the calculation process clearly, including how you chose the METs value based on the running speed.
3. If the exact METs value is difficult to determine, provide an estimate using standard METs values (e.g., 8 km/h = 8.3 METs), and explain accordingly.

In your response, include:
- The total calories burned
- A step-by-step breakdown of the calculation
- A friendly and easy-to-understand explanation
**Finally, provide your full answer in Korean.**

'''
                )
            },
            {
                "role":"user",
                "content":"시간 60분, 거리 10킬로, 체중 75, 나이 50, 성별 남성"
            }
    ]
)

print(response.choices[0].message.content)
```
